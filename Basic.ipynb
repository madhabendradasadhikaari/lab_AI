```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
import cv2
from sklearn.preprocessing import LabelEncoder
print("="*60)
print("INTRODUCTION TO PYTHON LIBRARIES FOR AI")
print("="*60)

#NUPY

print("\n------NumPy------")
numbers = np.array([10,20,30,40,50])
print("Array :",numbers)
print("Sum :",np.sum(numbers))
print("Mean :",np.mean(numbers))
print("Maximum :",np.max(numbers))

#PANDAS

print("\n------Pandas------")
student = {
    "Name":["Alice","Bob","Charlie"],
    "Marks":[85,90,78]
}
df=pd.DataFrame(student)
print("\nStundent Data")
print(df)

#MATPLOTLIB

print("\n------Matplotlib------")
print("Displaying Bar Chart...")
plt.figure(figsize=(6,4))
plt.bar(df["Name"],df["Marks"])
plt.title("Student Marks")
plt.xlabel("Student Marks")
plt.ylabel("Student Name")
plt.ylabel("Marks")
plt.show()

#SCIKIT-LEARN

print("\n------Scikit-learn------")
encoder = LabelEncoder()
df["Student ID"] = encoder.fit_transform(df["Name"])
print("After Encoding Names into Numbers")
print(df)

#TENSORFLOW

print("\n------TensorFlow------")
tensor1=tf.constant([1,2,3])
tensor2=tf.constant([4,5,6])
print("Tensor 1:",tensor1.numpy())
print("Tnesor 2:",tensor2.numpy())
result = tf.add(tensor1,tensor2)
print("Tensor Addition :", result.numpy())

#OpenCV

print("\n------OpenCV------")
#Create a blank image
image = np.zeros((250,250,3),dtype=np.uint8)
#Draw a blue rectangle
cv2.rectangle(image,(50,50),(200,200),(255,0,0),3)
#Draw a green circle
cv2.circle(image,(125,125),40,(0,255,0),3)
#Display image
plt.imshow(cv2.cvtColor(image,cv2.COLOR_BGR2RGB))
plt.title("OpenCV Shapes")
plt.axis("off")
plt.show()

#CONCLUSION

print("\n"+"="*60)
print("Experiment Completed Successfully")
print("="*60)
print("\nLibraries Demonstrated:")
print("NumPy")
print("Pandas")
print("Matplotlib")
print("Scikit-learn")
print("TensorFlow")
print("OpenCV")
print("")
```
