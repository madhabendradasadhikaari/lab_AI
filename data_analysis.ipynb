```python
!pip install -q kagglehub[pandas-datasets]

import kagglehub
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns

dataset_path = kagglehub.dataset_download(
    "kristoft/pitt-quantum-repository-106066-molecules"
)

print("Dataset downloaded successfully!")
print("Dataset location:")
print(dataset_path)

print("\nFiles available in the dataset:")

for root, directories, files in os.walk(dataset_path):
    for file in files:
        print(os.path.join(root, file))

csv_files = []

for root, directories, files in os.walk(dataset_path):
    for file in files:
        if file.endswith(".csv"):
            csv_files.append(os.path.join(root, file))

print("\nCSV files found:")

for file in csv_files:
    print(file)

if len(csv_files) > 0:
    file_path = csv_files[0]
    df = pd.read_csv(file_path)
    print("\nDataset loaded successfully!")
else:
    print("\nNo CSV file was found.")
    df = None

if df is not None:
    print("\nFirst 5 records:")
    print(df.head())

if df is not None:
    print("\nLast 5 records:")
    print(df.tail())

if df is not None:
    print("\nDataset Shape:")
    print(df.shape)

    print("\nNumber of rows:", df.shape[0])
    print("Number of columns:", df.shape[1])

if df is not None:
    print("\nColumn Names:")
    for column in df.columns:
        print(column)

if df is not None:
    print("\nDataset Information:")
    df.info()

if df is not None:
    print("\nStatistical Summary:")
    print(df.describe(include="all"))

if df is not None:
    print("\nMissing Values:")
    print(df.isnull().sum())

if df is not None:
    print("\nNumber of duplicate records:")
    print(df.duplicated().sum())

if df is not None:
    df = df.drop_duplicates()
    print("\nDataset shape after removing duplicates:")
    print(df.shape)

if df is not None:
    numerical_columns = df.select_dtypes(
        include=["int64", "float64"]
    ).columns

    print("\nNumerical Columns:")
    for column in numerical_columns:
        print(column)

if df is not None:
    categorical_columns = df.select_dtypes(
        include=["object", "category"]
    ).columns

    print("\nCategorical Columns:")
    for column in categorical_columns:
        print(column)

if df is not None:
    numerical_columns = df.select_dtypes(
        include=["int64", "float64"]
    ).columns

    for column in numerical_columns[:5]:
        plt.figure(figsize=(8, 5))
        plt.hist(df[column].dropna(), bins=30)
        plt.title("Distribution of " + column)
        plt.xlabel(column)
        plt.ylabel("Frequency")
        plt.show()

if df is not None:
    numerical_df = df.select_dtypes(
        include=["int64", "float64"]
    )

    if numerical_df.shape[1] > 1:
        plt.figure(figsize=(12, 8))
        sns.heatmap(
            numerical_df.corr(),
            annot=True,
            cmap="coolwarm"
        )
        plt.title("Correlation Between Numerical Features")
        plt.show()

if df is not None:
    print("\n======================================")
    print("FINAL DATASET SUMMARY")
    print("======================================")

    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])

    print("\nColumns:")
    print(list(df.columns))

    print("\nMissing values:")
    print(df.isnull().sum())

    print("\nDuplicate records:")
    print(df.duplicated().sum())
```
